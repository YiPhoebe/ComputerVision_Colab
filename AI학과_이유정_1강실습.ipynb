# AI영상데이터처리_실습_1강

## 파이썬의 자료구조의 차이점

| 자료구조 | 작성 방법 | 순서 | 중복 | 변경 가능 여부 | 활용 예 |
| --- | --- | --- | --- | --- | --- |
| 리스트 | `[값1, 값2]` | 유지 | 허용 | 요소 수정·추가·삭제 가능 | 색상 목록 |
| 튜플 | `(값1, 값2)` | 유지 | 허용 | 요소 교체·추가·삭제 불가 | 이미지 크기 |
| 딕셔너리 | `{키: 값}` | 삽입 순서 유지 | 키 중복 불가, 값 중복 가능 | 값 수정·항목 추가·삭제 가능 | 학생 정보 |
| 셋 | `{값1, 값2}` | 순서 보장 없음 | 허용하지 않음 | 원소 추가·삭제 가능 | 중복 제거, 집합 연산 |


순서가 있는 값을 수정하려면 리스트, 고정된 값 묶음은 튜플, 키로 정보를 찾으려면 딕셔너리, 중복을 제거하거나 집합을 비교하려면 셋을 사용합니다.


## Python의 4가지 자료 구조 예시

### 1. 리스트(List)
- 여러 값을 순차적으로 저장하는 데이터 타입으로, [] 대괄호를 사용하여 생성합니다.
- 리스트는 가변적으로 값을 수정, 추가, 삭제 할 수 있습니다.

```python
colors = ["red", "blue", "red"]

print("처음:", colors)
print("첫 번째 값:", colors[0])
print("red의 개수:", colors.count("red"))

colors[1] = "green"  # 인덱스로 수정
colors.append("yellow")  # 마지막에 추가
colors.remove("red")  # 첫 번째 red를 삭제

print("수정·추가·삭제 후:", colors)
print("앞의 두 값:", colors[:2])
```

### 2. 튜플(Tuple)
- 순서가 있고 중복을 허용하지만 생성 후 값을 변경할 수 없는 불변 데이터 타입으로, () 소괄호를 사용하여 생성합니다.
- 요소의 값을 변경할 수 없어서 데이터의 불변성을 유지해야할 때 사용합니다. (프로그램 안정성을 높이는데 도움이 됩니다.)
- 고정된 데이터를 저장하는데 유용하며, 반복적인 값이나 상수의 집합을 나타낼때 자주 사용됩니다.

```python
dimensions = (1920, 1080, 1920)

print("튜플:", dimensions)
print("첫 번째 값:", dimensions[0])
print("1920의 개수:", dimensions.count(1920))

try:
    dimensions[0] = 1280
except TypeError:
    print("TypeError: 튜플의 요소는 변경할 수 없습니다.")

print("한 요소 튜플:", (1920,))  # 쉼표가 필요합니다.
```

### 3. 딕셔너리(Dictionary)
- 키-값 쌍으로 데이터를 저장하는 데이터 타입으로, {} 중괄호를 사용하여 생성합니다.
- 각 키는 중복될 수 없으며, 키를 통해 값을 참조하고 수정할 수 있습니다.

```python
student = {"name": "Alice", "age": 22, "major": "Computer Science"}

print("처음:", student)
print("이름 조회:", student["name"])

student["age"] = 23  # 같은 키에 대입하면 기존 값이 바뀝니다.
student["grade"] = 3  # 새로운 키-값 쌍을 추가합니다.

del student["major"]  # 키-값 쌍을 삭제합니다.

print("수정 후:", student)
print("키 개수:", len(student))
print("키 목록:", list(student.keys()))
```

### 4. 셋(Set)
- 중복을 허용하지 않는 순서가 없는 데이터 타입으로, {} 중괄호를 사용하여 생성합니다.
- 주로 중복 요소를 제거하거나 교집합, 차집합 등의 집합 연산을 수행할 때 사용합니다.
- 집합은 수학적인 집합 개념을 반영하여, 중복된 값을 자동으로 제거하고 순서와 상관없이 데이터를 저장합니다.

```python
colors = {"red", "blue", "red"}

print("중복 제거:", colors)
print("원소 개수:", len(colors))  # red가 중복되어도 2개입니다.

colors.add("green")
colors.remove("blue")

print("추가·삭제 후:", colors)
print("red 포함 여부:", "red" in colors)

a = {1, 2, 3}
b = {3, 4, 5}

print("합집합:", sorted(a | b))
print("교집합:", sorted(a & b))
print("차집합:", sorted(a - b))
print("빈 셋:", set())  # {}는 빈 딕셔너리입니다.
```
